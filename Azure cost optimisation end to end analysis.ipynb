```python
df = df.rename(columns={
    'UsageDate': 'Usage_Date',
    'SubscriptionId': 'Subscription_Id',
    'SubscriptionName': 'Subscription_Name',
    'ResourceGroup': 'Resource_Group',
    'ResourceId': 'Resource_Id',
    'ServiceName': 'Service_Name',
    'MeterSubCategory': 'Meter_SubCategory',
    'ResourceLocation': 'Resource_Location',
    'UsageQuantity': 'Usage_Quantity',
    'UnitOfMeasure': 'Unit_Of_Measure',
    'PricingModel': 'Pricing_Model'
})
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df['Service_Name'] = df['Service_Name'].fillna(df['Service_Name'].mode()[0])
df['Meter_SubCategory'] = df['Meter_SubCategory'].fillna(df['Meter_SubCategory'].mode()[0])
df['Subscription_Name'] = df['Subscription_Name'].fillna(df['Subscription_Name'].mode()[0])
df['Resource_Location'] = df['Resource_Location'].fillna(df['Resource_Location'].mode()[0])
df['Unit_Of_Measure'] = df['Unit_Of_Measure'].fillna(df['Unit_Of_Measure'].mode()[0])
df['Currency'] = df['Currency'].fillna(df['Currency'].mode()[0])
df['Pricing_Model'] = df['Pricing_Model'].fillna(df['Pricing_Model'].mode()[0])
df['Subscription_Id'] = df['Subscription_Id'].fillna('Unknown')
df['Resource_Id'] = df['Resource_Id'].fillna('Unknown')
df['Resource_Group'] = df['Resource_Group'].fillna('Unknown')
df['Tags'] = df['Tags'].fillna(df['Tags'].mode()[0])
df['Usage_Quantity'] = df['Usage_Quantity'].fillna(df['Usage_Quantity'].median())
df['Cost'] = df['Cost'].fillna(df['Cost'].median())

```

```python
df = df.drop_duplicates()
```

```python
df['Resource_Location'] = df['Resource_Location'].replace({
    'westeurope': 'West Europe',
    'southeastasia': 'South Asia',
    'eastus2': 'East US',
    'eastus': 'East US',
    'northeurope': 'North Europe'
})
```

```python
df['Service_Name'] = df['Service_Name'].str.strip()
```

```python
df['Service_Name'] = df['Service_Name'].str.title()
```

```python
service_mapping = {
    'Cosmos Db': 'Cosmos DB',
    'Key Vault': 'Key Vault',
    'Sql Database': 'SQL Database',
    'App Service': 'App Service',
    'Azure Kubernetes Service': 'Azure Kubernetes Service',
    'Cognitive Services': 'Cognitive Services',
    'Networking': 'Networking',
    'Storage': 'Storage',
    'Backup': 'Backup'
}

df['Service_Name'] = df['Service_Name'].replace(service_mapping)
```
